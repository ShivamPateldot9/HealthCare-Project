```python
df.columns = df.columns.str.lower()
```

```python
df = df.rename(columns={
    'patientid': 'patient_id',
    'appointmentid': 'appointment_id',
    'scheduledday': 'scheduled_day',
    'appointmentday': 'appointment_day',
    'hipertension': 'hypertension',
    'handcap': 'handicap'
})
```

```python
df = df.rename(columns={'no-show' : 'no_show'})
```

```python
df.columns.tolist()
```

```python
df.dtypes
```

```python
df['patient_id'] = df['patient_id'].astype('int64')

df['scheduled_day'] = pd.to_datetime(df['scheduled_day']).dt.tz_localize(None)

df['appointment_day'] = pd.to_datetime(df['appointment_day']).dt.tz_localize(None)
```

```python
df.dtypes
```

```python
df.head(7)
```

```python
df['no_show_flag'] = (df['no_show'] == 'Yes').astype('int')
```

```python
df['no_show_flag'].mean()
```

```python
df = df[df['age'] >= 0]
```

```python
df.to_csv("cleaned_data.csv", index=False)
```
